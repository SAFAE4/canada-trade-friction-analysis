```python
import os
import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine, text

load_dotenv()
print("USER:", os.getenv("DB_USER"), "| DB:", os.getenv("DB_NAME"))
```

```python
url = (
    f"mysql+pymysql://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}"
    f"@{os.getenv('DB_HOST')}/{os.getenv('DB_NAME')}"
)
engine = create_engine(url)

with engine.connect() as conn:
    print(conn.execute(text("SELECT DATABASE(), CURRENT_USER()")).fetchall())
```

```python
df = pd.DataFrame({"province": ["Quebec", "Ontario"], "test_value": [1, 2]})
df.to_sql("setup_test", engine, if_exists="replace", index=False)

result = pd.read_sql("SELECT * FROM setup_test", engine)
print(result)
print("Rows written:", len(df), "| Rows read back:", len(result))
```
